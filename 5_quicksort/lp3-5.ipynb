```python
import random
import time


def deterministic_quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[-1]

    left = [x for x in arr[:-1] if x <= pivot]
    right = [x for x in arr[:-1] if x > pivot]

    return deterministic_quick_sort(left) + [pivot] + deterministic_quick_sort(right)


def randomized_quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot_index = random.randint(0, len(arr) - 1)
    pivot = arr[pivot_index]

    remaining = arr[:pivot_index] + arr[pivot_index + 1:]

    left = [x for x in remaining if x <= pivot]
    right = [x for x in remaining if x > pivot]

    return randomized_quick_sort(left) + [pivot] + randomized_quick_sort(right)


arr = list(map(int, input("Enter numbers separated by spaces: ").split()))

start = time.perf_counter()
sorted_det = deterministic_quick_sort(arr.copy())
end = time.perf_counter()

print("\nDeterministic Quick Sort:")
print("Sorted Array:", sorted_det)
print("Execution Time:", round(end - start, 8), "seconds")

start = time.perf_counter()
sorted_rand = randomized_quick_sort(arr.copy())
end = time.perf_counter()

print("\nRandomized Quick Sort:")
print("Sorted Array:", sorted_rand)
print("Execution Time:", round(end - start, 8), "seconds")
```
