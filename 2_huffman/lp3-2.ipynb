```python
import heapq
from collections import Counter

class Node:
    def __init__(self, char, freq):
        self.char = char
        self.freq = freq
        self.left = None
        self.right = None

    def __lt__(self, other):
        return self.freq < other.freq


def generate_codes(root, code, huffman_codes):
    if root is None:
        return

    if root.char is not None:
        huffman_codes[root.char] = code if code else "0"
        return

    generate_codes(root.left, code + "0", huffman_codes)
    generate_codes(root.right, code + "1", huffman_codes)


def huffman_encoding(text):
    if not text:
        print("Please enter a non-empty string.")
        return

    frequency = Counter(text)
    heap = []

    for char, freq in frequency.items():
        heapq.heappush(heap, Node(char, freq))

    while len(heap) > 1:
        left = heapq.heappop(heap)
        right = heapq.heappop(heap)

        new_node = Node(None, left.freq + right.freq)
        new_node.left = left
        new_node.right = right

        heapq.heappush(heap, new_node)

    root = heapq.heappop(heap)
    huffman_codes = {}

    generate_codes(root, "", huffman_codes)

    print("\nCharacter Frequencies and Huffman Codes:")
    for char, freq in sorted(frequency.items()):
        display_char = "Space" if char == " " else char
        print(f"{display_char}: Frequency = {freq}, Code = {huffman_codes[char]}")

    encoded_text = "".join(huffman_codes[char] for char in text)

    print("\nOriginal Text:", text)
    print("Encoded Text:", encoded_text)
    print("Original Size:", len(text) * 8, "bits")
    print("Encoded Size:", len(encoded_text), "bits")

    average_length = len(encoded_text) / len(text)
    print("Average Code Length:", round(average_length, 2), "bits/character")


text = input("Enter a string to encode: ")
huffman_encoding(text)
```
