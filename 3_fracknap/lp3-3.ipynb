```python
class Item:
    def __init__(self, value, weight):
        self.value = value
        self.weight = weight


def fractional_knapsack(capacity, items):
    items.sort(key=lambda item: item.value / item.weight, reverse=True)

    total_value = 0.0

    print("\nItems selected:")

    for item in items:
        if capacity == 0:
            break

        if item.weight <= capacity:
            capacity -= item.weight
            total_value += item.value

            print("Value:", item.value,
                  "Weight:", item.weight,
                  "Fraction taken: 1")

        else:
            fraction = capacity / item.weight
            total_value += item.value * fraction

            print("Value:", item.value,
                  "Weight:", item.weight,
                  "Fraction taken:", round(fraction, 2))

            capacity = 0

    return total_value


n = int(input("Enter the number of items: "))
items = []

for i in range(n):
    value = float(input("Enter value of item " + str(i + 1) + ": "))
    weight = float(input("Enter weight of item " + str(i + 1) + ": "))

    if weight <= 0 or value < 0:
        print("Weight must be positive and value cannot be negative.")
        exit()

    items.append(Item(value, weight))

capacity = float(input("Enter knapsack capacity: "))

if capacity < 0:
    print("Capacity cannot be negative.")
else:
    max_value = fractional_knapsack(capacity, items)
    print("\nMaximum value we can obtain:", round(max_value, 2))
```
